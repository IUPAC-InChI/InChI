```python
filename = "./data/cis_platin.mol"

test_mol = openbabel.OBMol()

# Create an OBConversion object
obConversion = openbabel.OBConversion()

# Set the input format ('mol') and output format ('xyz')
obConversion.SetInFormat("mol")
obConversion.SetOutFormat("xyz")

if not obConversion.ReadFile(test_mol, filename):
    print("Error: Could not read the mol file.")
else:
    xyz_string = obConversion.WriteString(test_mol)

    complex_mol = mol3D()
    complex_mol.readfromstring(xyz_string)
    print(complex_mol.get_geometry_type())
```

```python
directory = './data'

absolute_file_paths = [os.path.abspath(os.path.join(directory, f)) for f in os.listdir(directory) if os.path.isfile(os.path.join(directory, f)) and "mol" in f and not "xyz" in f]

tmp_mol = openbabel.OBMol()

# Create an OBConversion object
obConversion = openbabel.OBConversion()

# Set the input format ('mol') and output format ('xyz')
obConversion.SetInFormat("mol")
obConversion.SetOutFormat("xyz")

for cur_fn in absolute_file_paths:
    
    complex_mol = mol3D()

    try:
        print("----------------------")
        print(cur_fn)
        
        if not obConversion.ReadFile(tmp_mol, filename):
            print("Error: Could not read the mol file.")
        else:        
            xyz_string = obConversion.WriteString(test_mol)
            
            complex_mol = mol3D()
            complex_mol.readfromstring(xyz_string)
            
            geo_dict = complex_mol.get_geometry_type()
            if "geometry" in geo_dict:
                print(geo_dict["geometry"])
    except Exception as ex:
        # pass
        print("error", str(ex))
        
```

```python
#works only with mol v2

directory = './data'

absolute_file_paths = [os.path.abspath(os.path.join(directory, f)) for f in os.listdir(directory) if os.path.isfile(os.path.join(directory, f)) and "mol" in f and not "xyz" in f]

for cur_fn in absolute_file_paths:
    
    complex_mol = mol3D()

    try:
        print("----------------------")
        print(cur_fn)
        complex_mol.readfrommol(cur_fn)
        geo_dict = complex_mol.get_geometry_type()
        if "geometry" in geo_dict:
            print(geo_dict["geometry"])
    except Exception as ex:
        print("error", str(ex))
```
